```python
# Method 3 CHANGE A cell | ResNet50 (CNN) | fraction=25% | 15 epochs | seed 42  [PathVQA]
# Single run for running on a separate account. Subsamples official TRAIN IMAGES
# to 25% (all QAs for chosen images); val/test fixed; vocab/answers from FULL train.
# Writes scaling_cell_cnn_25.csv into OUTPUT_DIR. You have CNN & MiT @100% already.

# Method 3 (ResNet50) - ResNet50 + Stacked Attention + LayerNorm  [PathVQA]
# Changes 1,2,4,6 + native stacked-attention overlay. Runs ONLY the ResNet50 backbone.

# ============================================================
# PATHVQA - LAYER NORM + QUANTIZATION (Target 65%)
# - Layer Normalization after fusion (Paper 7)
# - Optional 8-bit dynamic quantization (Papers 4,8,10)
# - EfficientNet-B3 / ResNet50, 512px, 2-hop attention
# ============================================================

# =========================
# 0. INSTALLS
# =========================
!pip -q install datasets timm nltk rouge-score scikit-learn

# =========================
# 1. IMPORTS
# =========================
import os, re, gc, random, string, hashlib, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from io import BytesIO
from PIL import Image
from tqdm.auto import tqdm
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
import torchvision.transforms.functional as TF
import timm
from datasets import load_dataset
from sklearn.metrics import accuracy_score
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer
import nltk

nltk.download("stopwords")
from nltk.corpus import stopwords
warnings.filterwarnings("ignore")

# =========================
# 2. SEED / DEVICE
# =========================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("DEVICE:", DEVICE)

# =========================
# 3. CONFIG (with quantization flag)
# =========================
DATASET_NAME = "flaviagiammarino/path-vqa"

IMAGE_SIZE = 512
BATCH_SIZE = 8
ACCUMULATION_STEPS = 2
NUM_WORKERS = 2
EPOCHS = 15
LR = 1e-4
WEIGHT_DECAY = 1e-4
GRAD_CLIP = 1.0

MAX_Q_LEN = 30
WORD_VOCAB_SIZE = 2200

EMBED_DIM = 300
LSTM_HIDDEN = 768
LSTM_NUM_LAYERS = 2
LSTM_DROPOUT = 0.4
BIDIRECTIONAL = True

ATT_HIDDEN = 1024
NUM_ATT_HOPS = 2

LABEL_SMOOTHING = 0.1

# Quantization: set to True to apply 8-bit dynamic quantization after training
USE_QUANTIZATION = False   # Change to True for inference memory saving

OUTPUT_DIR = "/kaggle/working/pathvqa_scale_cnn_25"
os.makedirs(OUTPUT_DIR, exist_ok=True)

STOP_WORDS = set(stopwords.words("english"))
SMOOTH_FN = SmoothingFunction().method1
ROUGE_SCORER = rouge_scorer.RougeScorer(["rouge1", "rougeL"], use_stemmer=True)

# Models
MODEL_SPECS = [
    ("ResNet50 (CNN)", "resnet50"),   # CHANGE A single-cell backbone
]

# =========================
# 4. TEXT NORMALIZATION
# =========================
# CHANGE 2: no stopword removal (it corrupted answers).
def _basic_clean(x):
    x = str(x).lower().strip()
    x = x.translate(str.maketrans("", "", string.punctuation))
    x = re.sub(r"\s+", " ", x)
    return x.strip()

def normalize_question(x, unk_token="<unk_q>"):
    x = _basic_clean(x); return x if x else unk_token

def normalize_answer(x, unk_token="<unk_a>"):
    x = _basic_clean(x); return x if x else unk_token

# CHANGE 6: question type from raw question + raw answer
_YESNO_LEAD = {"is","are","does","do","did","was","were","can","could","has","have","will","should","would","may","am"}
def get_question_type(question_raw, answer_raw):
    a = str(answer_raw).strip().lower()
    if a in ("yes","no"): return "yes/no"
    q = _basic_clean(question_raw); first = q.split()[0] if q.split() else ""
    if first in ("what","which","whose","name"): return "what"
    if first == "where": return "where"
    if first == "how": return "how"
    if first == "when": return "when"
    if first in _YESNO_LEAD: return "yes/no"
    return "other"

def tokenize_text(x):
    x = str(x).strip()
    toks = x.split()
    if len(toks) == 0:
        toks = ["<unk>"]
    return toks

# =========================
# 5. IMAGE HELPERS
# =========================
def image_to_pil(img_obj):
    if isinstance(img_obj, Image.Image):
        return img_obj.convert("RGB")
    if isinstance(img_obj, dict):
        if img_obj.get("bytes") is not None:
            return Image.open(BytesIO(img_obj["bytes"])).convert("RGB")
        if img_obj.get("path") is not None:
            return Image.open(img_obj["path"]).convert("RGB")
    raise ValueError("Unsupported image object format")

def stable_image_id(img_obj, fallback="unknown"):
    try:
        if isinstance(img_obj, dict):
            if img_obj.get("path") is not None:
                return str(img_obj["path"])
            if img_obj.get("bytes") is not None:
                return hashlib.md5(img_obj["bytes"]).hexdigest()
        if isinstance(img_obj, Image.Image):
            buf = BytesIO()
            img_obj.convert("RGB").save(buf, format="PNG")
            return hashlib.md5(buf.getvalue()).hexdigest()
    except Exception:
        return str(fallback)
    return str(fallback)

# =========================
# 6. LOAD DATASET
# =========================
dataset = load_dataset(DATASET_NAME)
all_split_names = list(dataset.keys())
base_cols = dataset[all_split_names[0]].column_names

def find_col(cols, candidates):
    for c in candidates:
        if c in cols:
            return c
    return None

IMAGE_COL = find_col(base_cols, ["image", "img"])
QUESTION_COL = find_col(base_cols, ["question", "query"])
ANSWER_COL = find_col(base_cols, ["answer", "answers"])
CATEGORY_COL = find_col(base_cols, ["answer_type", "question_type", "category"])
IMAGE_ID_COL = find_col(base_cols, ["image_id", "img_id", "id"])

if IMAGE_COL is None or QUESTION_COL is None or ANSWER_COL is None:
    raise ValueError("Required columns not found.")

# =========================
# 7. BUILD FULL DATAFRAME
# =========================
def build_full_df(dataset_dict):
    rows = []
    for split_name in dataset_dict.keys():
        hf_split = dataset_dict[split_name]
        for i in tqdm(range(len(hf_split)), desc=f"Reading {split_name}"):
            ex = hf_split[i]
            img_obj = ex[IMAGE_COL]
            image_id = str(ex[IMAGE_ID_COL]) if (IMAGE_ID_COL is not None and IMAGE_ID_COL in ex) else stable_image_id(img_obj, fallback=f"{split_name}_{i}")
            q_raw = str(ex[QUESTION_COL])
            a_raw = str(ex[ANSWER_COL])
            q_proc = normalize_question(q_raw)
            a_proc = normalize_answer(a_raw)
            rows.append({
                "source_split": split_name,
                "source_idx": i,
                "image_id": image_id,
                "question_raw": q_raw,
                "answer_raw": a_raw,
                "question_proc": q_proc,
                "answer_proc": a_proc,
                "question_type": get_question_type(q_raw, a_raw)   # CHANGE 6
            })
    return pd.DataFrame(rows)

full_df = build_full_df(dataset)
print("\nTotal rows:", len(full_df))
print("Total images:", full_df["image_id"].nunique())

# =========================
# 8. CHANGE 1: OFFICIAL SPLIT (use each row's own source_split label)
# =========================
train_df_full = full_df[full_df["source_split"] == "train"].reset_index(drop=True)
val_df   = full_df[full_df["source_split"] == "validation"].reset_index(drop=True)
test_df  = full_df[full_df["source_split"] == "test"].reset_index(drop=True)

print("\n===== OFFICIAL PATHVQA SPLIT =====")
print("Train images:", train_df_full["image_id"].nunique(), " Train QA:", len(train_df_full))
print("Val images  :", val_df["image_id"].nunique(),   " Val QA  :", len(val_df))
print("Test images :", test_df["image_id"].nunique(),  " Test QA :", len(test_df))

# ============================================================
# CHANGE A (single-cell): subsample OFFICIAL TRAIN IMAGES at FRACTION.
# Image-level (carry all QAs for chosen images) -> no image leakage.
# val/test stay FIXED. Vocab + answer space are built from the FULL train set
# below, so this cell is comparable to the other fractions/backbones.
# ============================================================
FRACTION = 0.25          # this notebook's data fraction
SCALE_SEED = 42                  # single seed
BACKBONE_TAG = "cnn"

def _subsample_train_by_image(full_df_, frac, seed):
    imgs = sorted(full_df_["image_id"].unique().tolist())
    rng = random.Random(seed); rng.shuffle(imgs)
    k = max(1, int(round(frac * len(imgs))))
    keep = set(imgs[:k])
    return full_df_[full_df_["image_id"].isin(keep)].reset_index(drop=True)

train_df = _subsample_train_by_image(train_df_full, FRACTION, SCALE_SEED)
TRAIN_IMAGES_USED = int(train_df["image_id"].nunique())
TRAIN_QA_USED = int(len(train_df))
print(f"\n===== CHANGE A SUBSAMPLE | fraction={FRACTION:.0%} | seed={SCALE_SEED} =====")
print(f"Used train images: {TRAIN_IMAGES_USED}  Used train QA: {TRAIN_QA_USED}")

# =========================
# 9. WORD VOCAB
# =========================
word_counter = Counter()
for text in train_df_full["question_proc"].tolist() + train_df_full["answer_proc"].tolist():
    word_counter.update(tokenize_text(text))
SPECIAL_TOKENS = ["<pad>", "<unk>"]
most_common_words = [w for w, _ in word_counter.most_common(WORD_VOCAB_SIZE - len(SPECIAL_TOKENS))]
itos = SPECIAL_TOKENS + most_common_words
stoi = {w: i for i, w in enumerate(itos)}
PAD_IDX = stoi["<pad>"]
UNK_IDX = stoi["<unk>"]
print("Word vocab size:", len(stoi))

def encode_question(text, max_len=MAX_Q_LEN):
    toks = tokenize_text(text)[:max_len]
    ids = [stoi.get(tok, UNK_IDX) for tok in toks]
    q_len = max(1, len(ids))
    if len(ids) < max_len:
        ids += [PAD_IDX] * (max_len - len(ids))
    return ids, q_len

# =========================
# 10. ANSWER SPACE
# =========================
train_answer_counts = Counter(train_df_full["answer_proc"].tolist())
answer_list = sorted(train_answer_counts.keys())
if "<unk_a>" not in answer_list:
    answer_list = answer_list + ["<unk_a>"]
answer2id = {a: i for i, a in enumerate(answer_list)}
id2answer = {i: a for a, i in answer2id.items()}
def map_answer_to_id(answer_text):
    return answer2id.get(answer_text, answer2id["<unk_a>"])
print("Num answer classes:", len(answer2id))

# =========================
# 11. TRANSFORMS
# =========================
class RandomAffinePaper:
    def __call__(self, img):
        angle = random.uniform(-5, 5)
        tx = int(random.uniform(-0.08, 0.08) * img.size[0])
        ty = int(random.uniform(-0.08, 0.08) * img.size[1])
        scale = random.uniform(0.92, 1.08)
        shear = random.uniform(-8, 8)
        return TF.affine(img, angle=angle, translate=(tx, ty), scale=scale, shear=shear)

# CHANGE 4: augmentation removed so preprocessing matches Methods 1 & 2 (fair comparison).
train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

eval_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# =========================
# 12. DATASET
# =========================
class PathVQADataset(Dataset):
    def __init__(self, dataset_dict, df, transform):
        self.dataset_dict = dataset_dict
        self.df = df.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        split_name = row["source_split"]
        source_idx = int(row["source_idx"])
        ex = self.dataset_dict[split_name][source_idx]

        img = image_to_pil(ex[IMAGE_COL])
        img = self.transform(img)

        q_raw = str(row["question_raw"])
        q_proc = str(row["question_proc"]).strip()
        if q_proc == "":
            q_proc = "<unk_q>"

        a_raw = str(row["answer_raw"])
        a_proc = str(row["answer_proc"]).strip()
        if a_proc == "":
            a_proc = "<unk_a>"

        q_ids, q_len = encode_question(q_proc, MAX_Q_LEN)
        label = map_answer_to_id(a_proc)

        return {
            "image": img,
            "question_ids": torch.tensor(q_ids, dtype=torch.long),
            "question_len": torch.tensor(q_len, dtype=torch.long),
            "label": torch.tensor(label, dtype=torch.long),
            "question_raw": q_raw,
            "question_proc": q_proc,
            "answer_raw": a_raw,
            "answer_proc": a_proc,
            "image_id": str(row["image_id"]),
            "question_type": str(row["question_type"])   # CHANGE 6
        }

train_ds = PathVQADataset(dataset, train_df, train_transform)
val_ds = PathVQADataset(dataset, val_df, eval_transform)
test_ds = PathVQADataset(dataset, test_df, eval_transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

print("Train dataset:", len(train_ds))
print("Val dataset  :", len(val_ds))
print("Test dataset :", len(test_ds))

# =========================
# 13. GLOVE LOADER (manual download)
# =========================
def load_glove_embeddings(stoi, embed_dim=300):
    emb = np.random.normal(0, 0.02, (len(stoi), embed_dim)).astype(np.float32)
    emb[PAD_IDX] = 0.0
    glove_zip = "/kaggle/working/glove.6B.zip"
    glove_file = f"/kaggle/working/glove.6B.{embed_dim}d.txt"
    if not os.path.exists(glove_file):
        print("Downloading GloVe...")
        !wget -q -P /kaggle/working/ https://nlp.stanford.edu/data/glove.6B.zip
        !unzip -q /kaggle/working/glove.6B.zip -d /kaggle/working/
    print("Loading GloVe...")
    glove = {}
    with open(glove_file, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.split()
            word = parts[0]
            vec = np.array(parts[1:], dtype=np.float32)
            glove[word] = vec
    hits = 0
    for word, idx in stoi.items():
        if word in glove:
            emb[idx] = glove[word]
            hits += 1
    print(f"GloVe hits: {hits}/{len(stoi)}")
    return torch.tensor(emb, dtype=torch.float32)

glove_matrix = None   # CHANGE 4: from-scratch embeddings everywhere (no GloVe)

# =========================
# 14. MODEL WITH LAYER NORMALIZATION
# =========================
class QuestionEncoderLSTM(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim, num_layers=2, dropout=0.4, bidirectional=True, pretrained_embedding=None):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=PAD_IDX)
        if pretrained_embedding is not None:
            self.embedding.weight.data.copy_(pretrained_embedding)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, num_layers, batch_first=True,
                            dropout=dropout if num_layers > 1 else 0.0, bidirectional=bidirectional)
        self.dropout = nn.Dropout(dropout)
        self.bidirectional = bidirectional

    def forward(self, q_ids, q_lens):
        x = self.embedding(q_ids)
        q_lens = torch.clamp(q_lens, min=1).cpu()
        packed = nn.utils.rnn.pack_padded_sequence(x, q_lens, batch_first=True, enforce_sorted=False)
        _, (h_n, _) = self.lstm(packed)
        if self.bidirectional:
            h = torch.cat((h_n[-2], h_n[-1]), dim=1)
        else:
            h = h_n[-1]
        return self.dropout(h)

class FeatureExtractor(nn.Module):
    def __init__(self, timm_model_name):
        super().__init__()
        self.backbone = timm.create_model(timm_model_name, pretrained=True, features_only=True, out_indices=[-1])
        self.num_channels = self.backbone.feature_info.channels()[-1]
    def forward(self, x):
        return self.backbone(x)[0]

class StackedAttention(nn.Module):
    def __init__(self, visual_dim, hidden_dim, att_dim):
        super().__init__()
        self.W_v = nn.Linear(visual_dim, att_dim)
        self.W_q = nn.Linear(hidden_dim, att_dim)
        self.W_att = nn.Linear(att_dim, 1)
    def forward(self, V, q):
        q_exp = self.W_q(q).unsqueeze(1)
        v_proj = self.W_v(V)
        scores = self.W_att(torch.tanh(v_proj + q_exp)).squeeze(-1)
        att = F.softmax(scores, dim=1)
        attended = torch.bmm(att.unsqueeze(1), V).squeeze(1)
        return attended, att

class PathVQAStackedAttention(nn.Module):
    def __init__(self, timm_model_name, vocab_size, num_answers, embedding_matrix, num_hops=2, att_hidden=1024):
        super().__init__()
        self.num_hops = num_hops
        self.image_encoder = FeatureExtractor(timm_model_name)
        visual_dim = self.image_encoder.num_channels
        q_hidden = LSTM_HIDDEN * 2 if BIDIRECTIONAL else LSTM_HIDDEN

        self.question_encoder = QuestionEncoderLSTM(vocab_size, EMBED_DIM, LSTM_HIDDEN,
                                                    LSTM_NUM_LAYERS, LSTM_DROPOUT, BIDIRECTIONAL, embedding_matrix)
        self.attentions = nn.ModuleList([StackedAttention(visual_dim, q_hidden, att_hidden) for _ in range(num_hops)])
        self.q_proj = nn.Linear(visual_dim, q_hidden)

        combined_dim = visual_dim + q_hidden
        # ---- LAYER NORMALIZATION ADDED HERE (after fusion) ----
        self.fusion_norm = nn.LayerNorm(combined_dim)
        # -------------------------------------------------------
        self.classifier = nn.Sequential(
            nn.Linear(combined_dim, 1024), nn.ReLU(), nn.Dropout(0.5),
            nn.Linear(1024, 512), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(512, num_answers)
        )

    def forward(self, images, q_ids, q_lens, return_attention=False):
        feat_map = self.image_encoder(images)
        B, C, H, W = feat_map.shape
        V = feat_map.view(B, C, H*W).permute(0,2,1)
        q = self.question_encoder(q_ids, q_lens)
        current_query = q
        for att_layer in self.attentions:
            attended, att_map = att_layer(V, current_query)
            current_query = self.q_proj(attended) + current_query
            current_query = F.normalize(current_query, dim=1)
        combined = torch.cat([attended, q], dim=1)
        # Apply layer normalization before classifier
        combined = self.fusion_norm(combined)
        logits = self.classifier(combined)
        if return_attention:
            return logits, att_map.view(B, H, W)
        return logits

# =========================
# 15. LABEL SMOOTHING LOSS
# =========================
class LabelSmoothingCrossEntropy(nn.Module):
    def __init__(self, smoothing=0.1):
        super().__init__()
        self.smoothing = smoothing
    def forward(self, pred, target):
        n_classes = pred.size(1)
        log_pred = F.log_softmax(pred, dim=1)
        with torch.no_grad():
            true_dist = torch.zeros_like(pred)
            true_dist.fill_(self.smoothing / (n_classes - 1))
            true_dist.scatter_(1, target.unsqueeze(1), 1 - self.smoothing)
        return torch.mean(torch.sum(-true_dist * log_pred, dim=1))

criterion = LabelSmoothingCrossEntropy(LABEL_SMOOTHING)

# =========================
# 16. METRICS
# =========================
def answer_token_f1(pred, gold):
    pred_t = pred.split()
    gold_t = gold.split()
    if not pred_t and not gold_t: return 1.0
    if not pred_t or not gold_t: return 0.0
    common = Counter(pred_t) & Counter(gold_t)
    num_same = sum(common.values())
    if num_same == 0: return 0.0
    p = num_same / len(pred_t)
    r = num_same / len(gold_t)
    return 2 * p * r / (p + r)

def bleu_n(pred, gold, n):
    pred_t = pred.split()
    gold_t = gold.split()
    weights = {1:(1,0,0,0), 2:(0.5,0.5,0,0), 3:(1/3,1/3,1/3,0)}[n]
    try:
        return sentence_bleu([gold_t], pred_t, weights=weights, smoothing_function=SMOOTH_FN)
    except:
        return 0.0

def rouge_scores(pred, gold):
    try:
        s = ROUGE_SCORER.score(gold, pred)
        return s["rouge1"].fmeasure, s["rougeL"].fmeasure
    except:
        return 0.0, 0.0

def compute_metrics_from_texts(golds, preds):
    yesno_idx = [i for i,g in enumerate(golds) if g in ["yes","no"]]
    open_idx = [i for i,g in enumerate(golds) if g not in ["yes","no"]]
    metrics = {}
    metrics["overall_exact_match_pct"] = np.mean([int(g==p) for g,p in zip(golds,preds)]) * 100.0 if golds else 0.0
    if yesno_idx:
        metrics["yes_no_accuracy_pct"] = accuracy_score([golds[i] for i in yesno_idx], [preds[i] for i in yesno_idx]) * 100.0
    else:
        metrics["yes_no_accuracy_pct"] = np.nan
    if open_idx:
        o_true = [golds[i] for i in open_idx]
        o_pred = [preds[i] for i in open_idx]
        metrics["open_exact_match_pct"] = np.mean([int(g==p) for g,p in zip(o_true,o_pred)]) * 100.0
        metrics["macro_f1_pct"] = np.mean([answer_token_f1(p,g) for g,p in zip(o_true,o_pred)]) * 100.0
        metrics["bleu1_pct"] = np.mean([bleu_n(p,g,1) for g,p in zip(o_true,o_pred)]) * 100.0
        metrics["bleu2_pct"] = np.mean([bleu_n(p,g,2) for g,p in zip(o_true,o_pred)]) * 100.0
        metrics["bleu3_pct"] = np.mean([bleu_n(p,g,3) for g,p in zip(o_true,o_pred)]) * 100.0
        r1,rl = zip(*[rouge_scores(p,g) for g,p in zip(o_true,o_pred)])
        metrics["rouge1_f_pct"] = np.mean(r1) * 100.0
        metrics["rougeL_f_pct"] = np.mean(rl) * 100.0
    else:
        for k in ["open_exact_match_pct","macro_f1_pct","bleu1_pct","bleu2_pct","bleu3_pct","rouge1_f_pct","rougeL_f_pct"]:
            metrics[k] = np.nan
    metrics["total_samples"] = len(golds)
    return metrics

# =========================
# 17. TRAIN / EVAL with gradient accumulation
# =========================
def train_one_epoch(model, loader, optimizer, scheduler=None):
    model.train()
    running_loss = 0.0
    optimizer.zero_grad()
    for batch_idx, batch in enumerate(tqdm(loader, desc="Train", leave=False)):
        images = batch["image"].to(DEVICE)
        q_ids = batch["question_ids"].to(DEVICE)
        q_lens = batch["question_len"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        logits = model(images, q_ids, q_lens)
        loss = criterion(logits, labels)
        loss = loss / ACCUMULATION_STEPS
        loss.backward()

        if (batch_idx + 1) % ACCUMULATION_STEPS == 0:
            torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
            optimizer.step()
            if scheduler is not None:
                scheduler.step()
            optimizer.zero_grad()

        running_loss += loss.item() * labels.size(0) * ACCUMULATION_STEPS
    return running_loss / len(loader.dataset)

@torch.no_grad()
def evaluate(model, loader, save_csv_path=None):
    model.eval()
    running_loss = 0.0
    all_gold = []
    all_pred = []
    all_types = []   # CHANGE 6
    rows = []

    for batch in tqdm(loader, desc="Eval", leave=False):
        images = batch["image"].to(DEVICE)
        q_ids = batch["question_ids"].to(DEVICE)
        q_lens = batch["question_len"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        logits = model(images, q_ids, q_lens)
        loss = criterion(logits, labels)
        running_loss += loss.item() * labels.size(0)

        pred_ids = torch.argmax(logits, dim=1).cpu().tolist()
        pred_answers = [id2answer.get(p, "<missing>") for p in pred_ids]
        gold_answers = [str(g).strip() if str(g).strip() != "" else "<blank>" for g in batch["answer_proc"]]

        all_gold.extend(gold_answers)
        all_pred.extend(pred_answers)
        all_types.extend(list(batch["question_type"]))   # CHANGE 6

        for i in range(len(pred_answers)):
            rows.append({
                "image_id": batch["image_id"][i],
                "question_type": batch["question_type"][i],
                "question_raw": batch["question_raw"][i],
                "question_proc": batch["question_proc"][i],
                "gold_answer": gold_answers[i],
                "pred_answer": pred_answers[i],
                "correct": int(gold_answers[i] == pred_answers[i])
            })

    metrics = compute_metrics_from_texts(all_gold, all_pred)
    metrics["loss"] = float(running_loss / len(loader.dataset))
    # CHANGE 6: per-question-type exact match
    order = ["yes/no","what","where","how","when","other"]; buckets = {}
    for g, pr, t in zip(all_gold, all_pred, all_types):
        buckets.setdefault(t, []).append(int(g == pr))
    per_type = {}
    for t in order:
        if t in buckets and buckets[t]:
            per_type[t] = (float(np.mean(buckets[t])*100), len(buckets[t]))
    for t, hits in buckets.items():
        if t not in per_type:
            per_type[t] = (float(np.mean(hits)*100), len(hits))
    metrics["per_type"] = per_type
    if save_csv_path is not None:
        pd.DataFrame(rows).to_csv(save_csv_path, index=False)
    return metrics, pd.DataFrame(rows)

# =========================
# 18. VISUALIZATION
# =========================
@torch.no_grad()
def print_and_plot_10_predictions(model, dataset_obj, model_name, num_samples=10):
    model.eval()
    indices = list(range(len(dataset_obj)))
    random.Random(SEED).shuffle(indices)
    indices = indices[:min(num_samples, len(indices))]

    print("\n" + "#" * 120)
    print(f"10 TEST PREDICTIONS FOR {model_name} (Stacked Attention + LayerNorm)")
    print("#" * 120)

    for i, idx in enumerate(indices, 1):
        sample = dataset_obj[idx]
        image = sample["image"].unsqueeze(0).to(DEVICE)
        q_ids = sample["question_ids"].unsqueeze(0).to(DEVICE)
        q_len = sample["question_len"].unsqueeze(0).to(DEVICE)

        logits, att_map = model(image, q_ids, q_len, return_attention=True)
        pred_id = torch.argmax(logits, dim=1).item()
        pred_answer = id2answer.get(pred_id, "<missing>")
        if not isinstance(pred_answer, str) or pred_answer.strip() == "":
            pred_answer = "<blank>"
        gt_answer = str(sample["answer_proc"]).strip()
        if gt_answer == "":
            gt_answer = "<blank>"

        img_np = sample["image"].permute(1, 2, 0).cpu().numpy()
        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])
        img_np = std * img_np + mean
        img_np = np.clip(img_np, 0, 1)

        att_np = att_map[0].cpu().numpy()
        att_np = (att_np - att_np.min()) / (att_np.max() - att_np.min() + 1e-8)

        print("=" * 120)
        print(f"Sample {i}")
        print("Image ID     :", sample["image_id"])
        print("Q-type       :", sample["question_type"])
        print("Question Raw :", sample["question_raw"])
        print("Question Proc:", sample["question_proc"])
        print("Ground Truth :", repr(gt_answer))
        print("Prediction   :", repr(pred_answer))
        print("Correct      :", int(gt_answer == pred_answer))

        plt.figure(figsize=(12, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(img_np)
        plt.axis("off")
        plt.title("Original Image")

        plt.subplot(1, 2, 2)
        plt.imshow(img_np)
        plt.imshow(att_np, cmap="jet", alpha=0.5, extent=(0, img_np.shape[1], img_np.shape[0], 0))
        plt.axis("off")
        plt.title("Attention Overlay (last hop)")

        plt.suptitle(
            f"{model_name} (Stacked Attention + LayerNorm)\nQ: {sample['question_raw']}\nGT: {gt_answer} | Pred: {pred_answer}",
            fontsize=11
        )
        plt.tight_layout()
        plt.show()

# =========================
# 19. MAIN TRAIN LOOP with optional quantization
# =========================
all_results = []
failed_models = []

for display_name, timm_name in MODEL_SPECS:
    print("\n" + "=" * 140)
    print(f"RUNNING MODEL: {display_name}  |  TIMM: {timm_name}  |  Stacked Attention + LayerNorm")
    print("=" * 140)

    model_dir = os.path.join(OUTPUT_DIR, display_name.replace("/", "_"))
    os.makedirs(model_dir, exist_ok=True)

    try:
        model = PathVQAStackedAttention(
            timm_model_name=timm_name,
            vocab_size=len(stoi),
            num_answers=len(answer2id),
            embedding_matrix=glove_matrix,
            num_hops=NUM_ATT_HOPS,
            att_hidden=ATT_HIDDEN
        ).to(DEVICE)

        optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
        # CHANGE 4: LR scheduler removed so training matches Methods 1 & 2.

        best_monitor = -1.0
        best_path = os.path.join(model_dir, f"{display_name}_best.pt")
        history = []

        for epoch in range(1, EPOCHS + 1):
            print(f"\nEpoch {epoch}/{EPOCHS}")

            train_loss = train_one_epoch(model, train_loader, optimizer, scheduler=None)
            val_metrics, _ = evaluate(
                model,
                val_loader,
                save_csv_path=os.path.join(model_dir, f"{display_name}_val_preds_epoch{epoch}.csv")
            )

            monitor = val_metrics["overall_exact_match_pct"]

            epoch_row = {"epoch": epoch, "train_loss": train_loss}
            for k, v in val_metrics.items():
                if k == "per_type":
                    continue
                epoch_row[f"val_{k}"] = v
            history.append(epoch_row)

            current_lr = optimizer.param_groups[0]['lr']
            print(f"train_loss: {train_loss:.6f}  LR: {current_lr:.2e}")
            for k, v in val_metrics.items():
                if k == "per_type":
                    continue
                print(f"{k}: {v}")

            if monitor > best_monitor:
                best_monitor = monitor
                torch.save(model.state_dict(), best_path)
                print("Saved best model to:", best_path)

        pd.DataFrame(history).to_csv(os.path.join(model_dir, f"{display_name}_history.csv"), index=False)

        # Load best model
        model.load_state_dict(torch.load(best_path, map_location=DEVICE))
        
        # ----- OPTIONAL QUANTIZATION (for inference memory saving) -----
        if USE_QUANTIZATION:
            print("\nApplying 8-bit dynamic quantization...")
            # Move model to CPU for quantization
            model_cpu = model.to("cpu")
            # Quantize the model (dynamic quantization works well for LSTM and Linear layers)
            quantized_model = torch.quantization.quantize_dynamic(
                model_cpu, 
                {nn.Linear, nn.LSTM},   # Quantize these layer types
                dtype=torch.qint8
            )
            # Move back to device (note: quantized models are usually kept on CPU for inference)
            # For GPU, we would need to use torch.quantization.quantize, but dynamic quant is simpler.
            # We'll evaluate on CPU after quantization to avoid additional complexity.
            print("Quantization complete. Model will be evaluated on CPU.")
            eval_device = "cpu"
            model = quantized_model
        else:
            eval_device = DEVICE
            model = model.to(eval_device)
        
        # Evaluate on test set
        test_metrics, test_pred_df = evaluate(
            model,
            test_loader,
            save_csv_path=os.path.join(model_dir, f"{display_name}_test_predictions.csv")
        )

        print("\nFINAL TEST METRICS")
        for k, v in test_metrics.items():
            if k == "per_type":
                continue
            print(f"{k}: {v}")

        # CHANGE 6: per-question-type exact match
        print("\n" + "-"*60)
        print(f"PER-QUESTION-TYPE EXACT MATCH - {display_name}")
        print("-"*60)
        print(f"{'Question Type':<14}{'Exact Match (%)':>18}{'N':>10}")
        for qtype, (acc, n) in test_metrics.get("per_type", {}).items():
            print(f"{qtype:<14}{acc:>18.2f}{n:>10}")

        if not USE_QUANTIZATION:
            print_and_plot_10_predictions(model, test_ds, display_name, num_samples=10)
        else:
            print("Skipping attention visualization because quantized model is on CPU and attention map extraction may be slower; results are saved in CSV.")

        result_row = {"model": display_name, "timm_name": timm_name, "quantized": USE_QUANTIZATION}
        for k, v in test_metrics.items():
            if k == "per_type":
                continue
            result_row[k] = v
        for qtype, (acc, n) in test_metrics.get("per_type", {}).items():
            result_row[f"acc_{qtype.replace('/', '_')}"] = acc
        all_results.append(result_row)

        del model
        gc.collect()
        if torch.cuda.is_available() and not USE_QUANTIZATION:
            torch.cuda.empty_cache()

    except Exception as e:
        print(f"\nMODEL FAILED: {display_name}")
        print("Exception:", e)
        failed_models.append({"model": display_name, "reason": str(e)})
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        continue

# =========================
# 20. FINAL RESULTS SUMMARY
# =========================
results_df = pd.DataFrame(all_results)
if len(results_df) > 0:
    results_df = results_df.sort_values(by="overall_exact_match_pct", ascending=False).reset_index(drop=True)

failed_df = pd.DataFrame(failed_models)

print("\n" + "=" * 140)
print("FINAL RESULTS (Stacked Attention + LayerNorm + Optional Quantization)")
print("=" * 140)
print(results_df)

if len(failed_df) > 0:
    print("\nFAILED MODELS")
    print(failed_df)

results_df.to_csv(os.path.join(OUTPUT_DIR, "all_model_test_results.csv"), index=False)
failed_df.to_csv(os.path.join(OUTPUT_DIR, "failed_models.csv"), index=False)

if len(results_df) > 0 and not USE_QUANTIZATION:
    plt.figure(figsize=(14, 6))
    plt.bar(results_df["model"], results_df["overall_exact_match_pct"])
    plt.xticks(rotation=45, ha="right")
    plt.ylabel("Overall Exact Match (%)")
    plt.title("Model Comparison - Stacked Attention + LayerNorm")
    plt.tight_layout()
    plt.show()

print("\nArtifacts saved to:", OUTPUT_DIR)
print("Done.")


# =========================
# CHANGE A: WRITE COMPACT SCALING-CELL RESULT (for you to collect across accounts)
# =========================
CELL_TAG = f"{BACKBONE_TAG}_{int(round(FRACTION*100))}"
if len(results_df) > 0:
    r0 = results_df.iloc[0].to_dict()
    cell_row = {
        "backbone": r0.get("model"),
        "timm_name": r0.get("timm_name"),
        "fraction": FRACTION,
        "seed": SCALE_SEED,
        "train_images": TRAIN_IMAGES_USED,
        "train_qa": TRAIN_QA_USED,
        "overall_em": r0.get("overall_exact_match_pct"),
        "yes_no_acc": r0.get("yes_no_accuracy_pct"),
        "open_em": r0.get("open_exact_match_pct"),
        "macro_f1": r0.get("macro_f1_pct"),
    }
    out_csv = os.path.join(OUTPUT_DIR, f"scaling_cell_{CELL_TAG}.csv")
    pd.DataFrame([cell_row]).to_csv(out_csv, index=False)
    print("\nSaved scaling-cell result ->", out_csv)
    print(cell_row)
else:
    print("\nNo results produced; scaling-cell CSV not written (check for a failed run above).")
```
